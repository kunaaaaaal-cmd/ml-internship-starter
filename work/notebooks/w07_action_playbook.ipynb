## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review Checklist:
Before any action is taken based on the identified KPI drops or data drift, a human expert must review the following aspects to prevent incorrect conclusions or inappropriate automated responses:

1.  **Contextual Understanding:** Evaluate if there are any external factors (e.g., marketing campaigns, holidays, system outages, news events) that could explain the observed changes.
2.  **Data Quality & Integrity:** Verify that the data pipeline is functioning correctly, and there are no issues with data collection, processing, or logging that might be causing spurious alerts.
3.  **Definition Drift:** Confirm that the definitions of KPIs or features have not changed over time, which could lead to an apparent drop or drift without a true underlying issue.
4.  **Experiment Design:** If the KPI drop is related to an A/B test or experiment, confirm the experiment's design validity (e.g., proper randomization, sufficient sample size, correct metric calculation).
5.  **Segment Deep Dive:** Investigate if the KPI drop or data drift is localized to a specific segment of users, geography, platform, or time, to narrow down the potential root cause.
6.  **Historical Precedent:** Check if similar patterns have been observed previously and how they were resolved.

### The 'No-Go' List (What should never be automated or decided without human oversight):
Certain decisions and interpretations are inherently complex and require human judgment, context, and ethical consideration. These should *never* be fully automated or actioned solely based on algorithmic alerts:

1.  **Causal Claim Establishment:** While models can identify correlations, establishing true causality (e.g., 'X causes Y') requires carefully designed experiments (like A/B tests) and expert interpretation. Automated systems should not make causal claims.
2.  **Strategic Business Decisions:** Major strategic shifts, product launches, or policy changes based on data anomalies should always involve human strategists, product managers, and leadership. The playbook provides input, not final decisions.
3.  **Ethical Implications & Fairness:** Decisions affecting user experience, access, or fairness (e.g., changes to recommendation algorithms, content moderation policies) must undergo rigorous human ethical review, especially if data drift impacts sensitive demographic features.
4.  **Resource Allocation & Budgeting:** Allocating significant engineering or marketing resources to address an anomaly should be a human decision, weighing costs, benefits, and strategic priorities.
5.  **Interpreting Novel Anomalies:** For anomalies that don't fit known patterns or trigger predefined alerts, human creativity and expertise are essential to diagnose and respond effectively.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

### Monitoring and Retrain Triggers:
To ensure the continued validity and effectiveness of any models or recommendations derived from this playbook, continuous monitoring is crucial. The following triggers indicate that the underlying assumptions or model performance may have degraded, requiring a review, retraining, or recalibration of the models and the playbook's recommendations.

1.  **KPI Performance Degradation:**
    *   **Trigger:** Any KPI (e.g., conversion_rate, click_through_rate) shows a drop exceeding a predefined percentage (e.g., 5% to 10%) compared to a historical baseline or a control group that cannot be explained by external factors after human review.
    *   **Action:** Investigate the specific KPI, identify contributing factors, and potentially retrain or adjust models whose outputs influence this KPI.

2.  **Data Drift:**
    *   **Trigger:** Statistical tests (e.g., Kolmogorov-Smirnov test, Population Stability Index) indicate a significant change in the distribution of key input features (e.g., `user_age_group`, `region`, `traffic_source`) between the training data and production data, or between different time periods.
    *   **Action:** Analyze the drifted features to understand the shift, update data pipelines if schema changes occurred, and retrain models with fresh, representative data.

3.  **Model Performance Degradation (if applicable):**
    *   **Trigger:** Offline evaluation metrics (e.g., precision, recall, F1-score, RMSE, AUC) or online A/B test results for a deployed model fall below acceptable thresholds.
    *   **Action:** Re-evaluate model features, architecture, and training data; retrain the model; or consider alternative modeling approaches.

4.  **External Event Impact:**
    *   **Trigger:** Significant external events (e.g., major product launch, competitor action, economic downturn, policy changes) that are likely to alter user behavior or data patterns.
    *   **Action:** Proactively review existing models and recommendations for relevance and potential bias, even without explicit alerts, and retrain if necessary.

5.  **Feedback Loop Analysis:**
    *   **Trigger:** Regular reviews of human feedback, A/B test outcomes, or post-action analyses reveal that previous recommendations from the playbook consistently led to suboptimal or incorrect outcomes.
    *   **Action:** Revisit the logic of the playbook itself, update reason codes, and refine action prioritization strategies.
