## 5. Why ML beats a fixed rule here

Machine learning beats a fixed rule for predicting housing prices due to the **inherent complexity and non-linear relationships** within the data.

**What makes the pattern too messy for an if-statement?**
*   **Multifactorial Influence**: Housing prices are influenced by numerous factors simultaneously, such as median income, house age, number of rooms, population density, average occupancy, location (latitude/longitude), and more. A fixed rule would struggle to account for the interplay of all these variables.
*   **Non-linear Relationships**: The relationship between these features and house price is rarely linear. For example, the impact of an extra room might differ significantly based on the house's age or location. Simple if-else statements cannot capture these complex, non-linear interactions.
*   **Interaction Effects**: Features often interact in non-obvious ways. A house with many rooms in a high-income area might be priced very differently than a house with many rooms in a low-income area. Fixed rules would require an explosion of nested conditions to attempt to model such interactions, quickly becoming unmanageable and brittle.
*   **Generalization**: A fixed rule defined for one region or time period would likely perform poorly in another, whereas an ML model can learn patterns that generalize better across different contexts. ML models can adapt to nuanced patterns and capture subtle trends that would be impossible to hard-code.
