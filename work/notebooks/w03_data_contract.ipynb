## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

For a generic search intelligence dataset, fields can be categorized as follows:

*   **Features**: Inputs used to predict or understand an outcome. These describe the query or user context.
    *   `query_text`: The actual search query entered by the user.
    *   `device_type`: (e.g., 'mobile', 'desktop', 'tablet').
    *   `user_country`: Geographic location of the user.
    *   `search_results_count`: Number of results returned for the query.
    *   `query_length`: Number of characters or words in the `query_text`.
    *   `past_query_count_24h`: Number of queries made by the user in the last 24 hours.

*   **Label**: The target variable we are trying to predict or measure, often indicating user satisfaction or success.
    *   `has_clicked_result`: Boolean (true/false) indicating if the user clicked on any search result.
    *   `conversion_event`: Boolean (true/false) if a conversion (e.g., purchase, form submission) occurred after the search.
    *   `clicked_result_rank`: The rank of the first clicked search result (if any).

*   **Context**: Information that describes the environment or specific instance of the data, but is not directly a feature or label for a model.
    *   `query_event_id`: Unique identifier for each search query event.
    *   `user_id`: Anonymous identifier for the user.
    *   `session_id`: Identifier for the user's current session.
    *   `timestamp`: When the query event occurred.
    *   `search_page_variant`: A/B test variant for the search results page.

*   **Excluded**: Fields that will not be used in the analysis or modeling, with a clear reason.
    *   `full_ip_address`: **Reason**: Contains Personally Identifiable Information (PII) and is not needed for the analytical task; a hashed or anonymized version might be used if geographical or network information is critical.
    *   `raw_user_agent_string`: **Reason**: Often too verbose and contains redundant information; `device_type` (derived from this) is usually sufficient.
    *   `internal_debugging_logs`: **Reason**: Not relevant for understanding user behavior or model training; intended for system diagnostics.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Understanding the limitations of your data is crucial for interpreting results and making informed decisions. For a typical search intelligence dataset, common limitations might include:

*   **Unbalanced History**: Data might be heavily skewed towards recent events, or specific user segments due to system changes or data retention policies. This can lead to biases in models trained on this data.
*   **GSC-only early rows (if applicable)**: If data is sourced from Google Search Console (GSC) for historical periods, it might lack granularity (e.g., no `user_id`, `session_id`) or include aggregated data, making it incompatible with more detailed event-level data from other sources.
*   **Window Overlaps / Gaps**: Inconsistent data collection or processing pipelines can lead to overlapping data entries for the same event or gaps in the time series, affecting time-based analysis.
*   **Attribution Challenges**: It might be difficult to precisely attribute conversions or user actions solely to a search query, as users might interact with multiple channels before converting.
*   **Lack of Negative Samples**: Often, data captures what users *did* (e.g., clicked queries) but not what they *didn't do* (e.g., queries where they found no relevant results and abandoned), which is vital for some analyses.
*   **Sampling Bias**: If the data is sampled (e.g., due to volume), the sample might not be representative of the entire population.
*   **Limited User Information**: Privacy concerns often restrict the amount of identifiable user information available, making it challenging to build highly personalized models.
